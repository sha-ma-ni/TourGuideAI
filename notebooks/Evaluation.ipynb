#### Manuelle Evaluation der Ergebnisse

Groundedness: Alle Antworten basieren auf den bereitgestellten Dokumenten.

Faithfulness: Alle Antworten geben die Informationen aus den Dokumenten korrekt wieder, ohne Verzerrungen oder Fehlinterpretationen.

Answer Relevancy: Alle Antworten sind relevant für die gestellten Fragen und liefern die benötigten Informationen.

Skala: 1 = gut, 2 = mittel, 3 = schlecht
