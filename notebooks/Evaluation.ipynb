#### Manuelle Evaluation der Ergebnisse

Groundedness: Alle Antworten basieren auf den bereitgestellten Dokumenten.

Faithfulness: Alle Antworten geben die Informationen aus den Dokumenten korrekt wieder, ohne Verzerrungen oder Fehlinterpretationen.

Answer Relevancy: Alle Antworten sind relevant für die gestellten Fragen und liefern die benötigten Informationen.

Skala: 1 = gut, 2 = mittel, 3 = schlecht

Insgesamt ist die Qualität der Antworten stark von der Komplexität der Fragen abhängig. Es wurde nur eine Quelle (Wikipedia) benutzt, obwohl es in anderen Quellen (z.B. Probleme: mangelnde Quellensammlung, es wird ein besserer Prompt benötigt.

#### Evaluation mit automatischer Metriken: Precision@k, Recall@k




**1. Precision@k:** Misst den Anteil der relevanten Dokumente unter den Top-k zurückgegebenen Dokumenten.

Precision@k = (Anzahl der relevanten Dokumente unter den Top-k zurückgegebenen Dokumenten) / k
