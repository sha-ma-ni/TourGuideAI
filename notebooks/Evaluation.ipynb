#### Manuelle Evaluation der Ergebnisse

Groundedness: Alle Antworten basieren auf den bereitgestellten Dokumenten.

Faithfulness: Alle Antworten geben die Informationen aus den Dokumenten korrekt wieder, ohne Verzerrungen oder Fehlinterpretationen.

Answer Relevancy: Alle Antworten sind relevant für die gestellten Fragen und liefern die benötigten Informationen.

Skala: 1 = gut, 2 = mittel, 3 = schlecht

Fazit: Die Antworten sind größtenteils korrekt, wenn es um präzise Fragen geht, besonderes, wenn man nach Öffnungszeiten oder Preisen fragt. Bei komplexeren Fragen oder Fragen, die mehr Informationen für eine Antwort benötigen, kommt es jedoch zu Halluzinationen. Häufig wird Wikipedia als Quelle verwendet und viele Informationen werden erfunden, obwohl sie in den Dokumenten teilweise vorhanden sind. Insgesamt ist die Qualität der Antworten stark von der Komplexität der Fragen abhängig. Es wurde nur eine Quelle (Wikipedia) benutzt, obwohl es in anderen Quellen (z.B. OSM) weitere Informationen verfügbar sind.
Probleme: mangelnde Quellensammlung, es wird ein besserer Prompt benötigt.
